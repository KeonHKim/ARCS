# ARCS v0.6.3 · axial pipe length at runtime

A fresh full-Maxwell BEM sweep with 256 panels and 100 frequencies from 10 Hz to 10 THz, for a 5 m uniform circular copper pipe. The cross-section uses millimetres; the axial length parameter always uses metres. Total longitudinal impedance is in Ω and dipolar/quadrupolar coefficients are in Ω/m. Circular m=0/m=1 references are multiplied by the same length. Entrance/exit effects are excluded. The constant-conductivity model is a numerical benchmark throughout this band, not a physical material certification.

**Saved output provenance:** this 5 m calculation was executed with v0.6.2. Its length is explicitly selected, so the new v0.6.3 default of 1.0 m does not change this calculation.


````python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
from IPython.display import display, Image
root = Path.cwd()
if not (root / "src/arcs").exists():
    root = root.parent
sys.path.insert(0, str(root / "src"))
sys.path.insert(0, str(root / "examples"))
import arcs
print("ARCS", arcs.__version__, arcs.__file__)
assert arcs.__version__ == "0.6.6", "Install the new release and restart the kernel."
from pipe_length_benchmark import run
summary, figures = run(pipe_length_m=5.0, points=100)
print(summary)
for name, figure in figures.items():
    display(Image(filename=str(root / "results/pipe_length" / (name + ".png"))))
    plt.close(figure)
````

For your own saved geometry:
```python
from arcs import load_geometry
from arcs.plotting import plot_longitudinal, plot_transverse
case = load_geometry("my_pipe.json")
solver = case.make_solver(pipe_length_m=5.0)
z = solver.longitudinal(f)                 # Ω
response = solver.transverse(f)             # tensor coefficients: Ω/m
plot_longitudinal(f, z, per_unit_length=False)
plot_transverse(f, response)                 # units inferred from recorded length
```

Omitting `pipe_length_m` keeps the usual per-length values. To integrate an existing per-length transverse result once, call `response.for_pipe_length(5.0)`. References used in plots must have the same normalization. The longitudinal plotting option changes labels only.

From the terminal:
```bash
python examples/run_case.py my_pipe.json --pipe-length-m 5 --output results/pipe_5m
```

**Maximum panel length** controls cross-section segments. **Gauss points / panel** controls interior geometry samples for display/export, independent of pipe length and the Maxwell field quadrature.
