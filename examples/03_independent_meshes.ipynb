# ARCS v0.6.3 · independent meshes for material interfaces

This runs a new full-Maxwell BEM sweep at 100 frequencies from 10 Hz to 10¹³ Hz, using a saved aperture mesh with 128 panels and an independently saved coating boundary with 192. Black curves use independent cylindrical m=0/m=1 matching; open blue markers are BEM results. The 5 mm NEG-labelled layer is deliberately thick to display mesh independence; its DC material model is a numerical test, not a realistic coating specification.

**Saved output provenance:** these completed calculations were executed with ARCS v0.6.1. The code below runs with v0.6.3. Notebook 04 contains a fresh v0.6.3 calculation with axial pipe length.


```python
%matplotlib inline
from pathlib import Path
import sys
ROOT=Path.cwd()
if not (ROOT / "src/arcs").exists(): ROOT=ROOT.parent
sys.path.insert(0,str(ROOT / "src"))
sys.path.insert(0,str(ROOT / "examples"))
import arcs
import matplotlib.pyplot as plt
from IPython.display import display
from independent_mesh_benchmark import run
print("ARCS",arcs.__version__,arcs.__file__)
assert arcs.__version__=="0.6.6"
```

```python
summary, figures=run(points=100)
for figure in figures:
    display(figure)
    plt.close(figure)
```

To edit this case in the GUI, open `examples/geometries/independent_meshes.json`, choose **Mesh settings**, click either boundary edge, change its options in **Boundary mesh**, and click **Generate / update mesh**. Save case preserves the separate settings. **Use shared defaults** resets only the selected boundary. Very thin curved interfaces require strictly nested chord meshes; crossing partitions are rejected. Generated roughness interfaces retain the aperture partition.
